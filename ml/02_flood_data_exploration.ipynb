```python
import pandas as pd

flood_df = pd.read_csv("data/flood_inventory.csv")

print(flood_df.shape)
print(flood_df.columns.tolist())

display(flood_df.head())
```

```python
rainfall_df = pd.read_csv("data/rainfall.csv", sep=";")

print(rainfall_df.shape)
```

```python
flood_df["Start Date"] = pd.to_datetime(
    flood_df["Start Date"],
    dayfirst=True,
    errors="coerce"
)

flood_2022 = flood_df[
    flood_df["Start Date"].dt.year == 2022
].copy()

print("2022 flood events:", len(flood_2022))

display(
    flood_2022[
        ["Start Date", "State", "Districts",
         "Latitude", "Longitude", "Severity"]
    ].head(10)
)
```

```python
print("2022 flood records:", len(flood_2022))
print("District values:", flood_2022["Districts"].notna().sum())
```

```python
print(
    flood_2022["Districts"]
    .dropna()
    .head(30)
    .to_list()
)
```

```python
# Clean state and district names
flood_2022["State"] = flood_2022["State"].str.strip()
flood_2022["Districts"] = flood_2022["Districts"].str.strip()

# Keep only rows with a specific district
flood_specific = flood_2022[
    flood_2022["Districts"].notna() &
    ~flood_2022["Districts"].str.contains(
        "several other|multiple|other district",
        case=False,
        na=False
    )
].copy()

print("Usable flood records:", len(flood_specific))
```

```python
print(flood_specific[["State", "Districts"]].head(30))
```

```python
# Keep only the first district when multiple districts are listed
flood_specific["District"] = (
    flood_specific["Districts"]
    .str.split(",")
    .str[0]
    .str.strip()
)

# Standardize text
flood_specific["State"] = flood_specific["State"].str.strip()
flood_specific["District"] = flood_specific["District"].str.strip()

print(flood_specific[["State", "District"]].head(20))
```

```python
print("Unique flood districts:", flood_specific["District"].nunique())
print("Unique states:", flood_specific["State"].nunique())
```

```python
rainfall_df["State"] = rainfall_df["state"].str.strip()
rainfall_df["District"] = rainfall_df["district"].str.strip()
```

```python
rainfall_districts = set(
    rainfall_df["District"].str.lower()
)

flood_districts = set(
    flood_specific["District"].str.lower()
)

matched = rainfall_districts.intersection(flood_districts)

print("Rainfall districts:", len(rainfall_districts))
print("Flood districts:", len(flood_districts))
print("Matched districts:", len(matched))
```

```python
day_columns = [
    "1st", "2nd", "3rd", "4th", "5th", "6th", "7th",
    "8th", "9th", "10th", "11th", "12th", "13th", "14th",
    "15th", "16th", "17th", "18th", "19th", "20th", "21st",
    "22nd", "23rd", "24th", "25th", "26th", "27th", "28th",
    "29th", "30th", "31st"
]

rainfall_daily = rainfall_df.melt(
    id_vars=["State", "District", "month"],
    value_vars=day_columns,
    var_name="Day",
    value_name="Rainfall"
)

rainfall_daily["Day"] = (
    rainfall_daily["Day"]
    .str.extract(r"(\d+)")
    .astype(int)
)

print(rainfall_daily.shape)
display(rainfall_daily.head())
```

```python
print(rainfall_daily.shape)
display(rainfall_daily.head())
```

```python
rainfall_daily["Date"] = pd.to_datetime(
    "2022-" +
    rainfall_daily["month"].astype(str) + "-" +
    rainfall_daily["Day"].astype(str),
    errors="coerce"
)

print(rainfall_daily.shape)
display(rainfall_daily.head())
```

```python
print(rainfall_daily["Date"].min())
print(rainfall_daily["Date"].max())
```

```python
rainfall_daily = rainfall_daily.dropna(subset=["Date"]).copy()

print(rainfall_daily.shape)
```

```python
flood_specific["Date"] = pd.to_datetime(
    flood_specific["Start Date"]
).dt.normalize()
```

```python
rainfall_daily["State"] = rainfall_daily["State"].str.strip().str.lower()
rainfall_daily["District"] = rainfall_daily["District"].str.strip().str.lower()

flood_specific["State"] = flood_specific["State"].str.strip().str.lower()
flood_specific["District"] = flood_specific["District"].str.strip().str.lower()
```

```python
flood_keys = flood_specific[
    ["State", "District", "Date"]
].drop_duplicates()

flood_keys["flood_occurred"] = 1
```

```python
ml_data = rainfall_daily.merge(
    flood_keys,
    on=["State", "District", "Date"],
    how="left"
)

ml_data["flood_occurred"] = (
    ml_data["flood_occurred"]
    .fillna(0)
    .astype(int)
)

print(ml_data.shape)
print(ml_data["flood_occurred"].value_counts())
```

```python
ml_data.to_csv("../data/processed/flood_rainfall_2022.csv", index=False)
```

Feature Engineering

```python
ml_data = ml_data.sort_values(
    ["State", "District", "Date"]
).copy()

ml_data["rainfall_3day"] = (
    ml_data.groupby(["State", "District"])["Rainfall"]
    .transform(lambda x: x.rolling(3, min_periods=1).sum())
)

ml_data["rainfall_7day"] = (
    ml_data.groupby(["State", "District"])["Rainfall"]
    .transform(lambda x: x.rolling(7, min_periods=1).sum())
)

print(ml_data.shape)
display(ml_data.head())
```

```python
ml_data.to_csv(
    "data/processed/flood_rainfall_features_2022.csv",
    index=False
)

print("Feature dataset saved successfully!")
```
